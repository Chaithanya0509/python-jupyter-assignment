### bitwise AND operation

```python
print(6&3,13&4)
```

### bitwise OR operation

```python
print(6|3,13|4)
```

### bitwise NOT operation

```python
print(~6,~19)
```

### bitwise XOR operation

```python
print(6^3,13^4)
```

```python
#left shift
print(53<<1,53<<2,53<<3,53<<4) #left shift by 1/2/3/4 bits
```

```python
#right shift
print(53>>1,53>>2,53>>3,53>>4) #right shift by 1/2/3/4 bits
```

## LIST:
#### it is represented with [],lists are mutable,indexing is applicable on lists.

```python
nums=[10,20,30,40,50]
print(nums,type(nums))
```

```python
print(dir(list))
```

```python
# append,insert,clear
nums=[10,20,30]
#append adds an element in the list at the end.
nums.append(50)
print(nums)
nums.append("Python")
print(nums)
nums.append(['a','b',25])
print(nums)
print(len(nums))
```

```python
info=[10, 20, 30, 40, 50, 'python', ['a', 'b', 25]]
#indexing and slicing(similar to string indexing and slicing)
print(info[2],info[5],info[-1])
print(info[2:])
print(info[2:4])
print(info[::2])
print(info[::-1])
```

```python
info=[10,20,30,50,'Python',['a','b',25]]
# multi-indexing
print(info[-1],type(info[-1]))
print(info[-1][-1],type(info[-1][-2]))
print(info[4],(info[4][3]))
```

```python
# insert and clear
# insert: to add an element at any given index.
nums=[10,20,30,40,50]
nums.insert(2,25)
print(nums)
nums.insert(0,2)
print(nums)
nums.insert(-1,200)
print(nums)
# to insert an element at the end,provide any index>=len(list)
nums.insert(12,200)
print(nums)

# clear:it clears the list,i.e it removes all the elements and keep an empty list
nums.clear()
print(nums)
```

```python
# extend:to append multiple elements in the lists:
nums1=[12,54,32,65,21]
nums2=[50,20,60,15]
#append all elements from nums2 in nums1 individually
#nums1.append(nums2)#this will insert nums2  into nums1 as the single elememt.
nums1.extend(nums2)
print(nums1,nums2)
```

```python
#copy:it is used to copy a list into another variable/location 
nums1=[10,20,30,40]
nums2=nums1#simple assingment copies the reference to the same list.
print(nums1,nums2)
nums2.append(50)
print(nums1,nums2)
```

```python
nums1=[10,20,30,40]
nums2=nums1.copy()#copy creates a separate list and stored in seperate variables
print(nums1,nums2)
nums2.append(50)
print(nums1,nums2)
```

```python
#count:to count appearances of a given element
nums=[10,20,40,20,20,10,50]
print(nums.count(10))
print(nums.count(20))
print(nums.count(30))
print(nums.count(40))
```

```python
#remove:to remove first appearances of a given element.
nums=[10,20,40,20,20,10,50]
nums.remove(10)
print(nums)
nums.remove(10)
print(nums)
#nums.remove(60)#value error:element not in the list
```

```python
# index:to get an index of first appearances of an element after given 'start' index:
nums=[10,20,40,20,20,10,50]
print(nums.index(20))
print(nums.index(50))
print(nums.index(20,2))#starting index for searching is 2
#print(nums.index(40,4))#value error:not in list
#print(nums.index(60))#value error:not in list

```

```python
#pop:it removes an element at given index(default=-1) and returns the same
nums=[10,20,30,40]
print(nums.pop())#40
print(nums)
print(nums.pop(1))#20
print(nums)
nums.pop()#30
nums.pop()#10
print(nums)
#nums.pop()#indexerror:pop empty list
```

```python
# sort:
nums=[234,54,65,23,435,230]
nums.sort()#by default in increasing order;original list is updated
print(True)
nums.sort(reverse=True)#sorting in decreasing order
print(nums)
```

### in-place functions:functions that changes the original list.in-place functions on list:'append','clear','extend','insert','pop','remove','reverse','sort'.

#### Tuple:it is represented with parameters(),it is immutable,indexing is applicable on tuple

```python
t=(10,30,5,10)
print(t,type(t))
#dir(tuple)#count,index
print(t.count(10))
print(t.index(5))

print(t[2])
#t[1]=100 #type error:'tuple' is immutable
```

### Set:set is a collection of unique objects,it is represented using{},it is mutable,it is unordred and so indexing is not applicable

```python
s={12,34,25,12,12,25,67,8,5}
print(s)
#s[2]# type error:'set' object id not subscriptable
```

```python
A={8,10,12,14}
B={7,8,15}
print(A,B)
print(A.intersection(B),B.intersection(A))
print(A.union(B),B.union(A))
print(A.difference(B),B.difference(A))
print(A-B,B-A)
```

```python
#add :used to add new element in set.
A={8,10,12,14}
A.add(80)
print(A)
A.add(12)#no error on adding duplicate;it simply gets ignored
print(A)
```

```python
#copy,clear,pop,remove:work on the set,similar to sets
s1={8,10,12,14}
s2={12,14}
s3={12,1,15,16}
s4={15,16}
print(s1.issubset(s2),s1.issuperset(s2))
print(s2.issubset(s1),s2.issuperset(s1))
print(s1.isdisjoint(s3),s1.isdisjoint(s4))
```

```python
#update and symmetric_updates
s1={8,10,12,12}
s2={12,14,15,16}
print(s1,s2)
s1.update(s2)
print(s1,s2)
```
