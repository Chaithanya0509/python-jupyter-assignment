### bitwise AND operation

```python
print(6&3,13&4)
```

### bitwise OR operation

```python
print(6|3,13|4)
```

### bitwise NOT operation

```python
print(~6,~19)
```

### bitwise XOR operation

```python
print(6^3,13^4)
```

```python
#left shift
print(53<<1,53<<2,53<<3,53<<4) #left shift by 1/2/3/4 bits
```

```python
#right shift
print(53>>1,53>>2,53>>3,53>>4) #right shift by 1/2/3/4 bits
```
