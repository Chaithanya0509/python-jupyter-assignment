### Q:
* get numbers input from the user until user enters 'q'(to quit).then print the list of all the numbers and their statistics like count,sum,min,max,average,etc

```python
nums=[]
while True:
  v=input("enter a number ('q' to quit):")
  if v=='q':
    break
  nums.append(float(v))
print("entered numbers:",nums)
print("total count:",len(nums))
print("total sums:",sum(nums))
print("minimum:",min(nums))
print("maximum:",max(nums))
print("average:",sum(nums)/len(nums))
```

### Q.
* build a calculator application to take an operator(+,-,/,//,%,^) and 2 input from the user and show the corresponding opeartion result.do this as long as user enter 'q' to quit

```python
while True:
  op = input("Enter an operator (+,-,*,/,//,%,^) ('q' to quit):")
  if op == 'q':
    break
  if op not in ['+','-','*','/','//','%','^']:
    print("Invlid operator!")
    continue
  n1 = float(input("Enter first number:"))
  n2 = float(input("Enter second number:"))
  result = None
  if op=="+":
    result = n1+n2
  elif op=="-":
    result = n1-n2
  elif op=="*":
    result = n1*n2
  elif op=="/":
    result = n1/n2
  elif op=="%":
    result = n1%n2
  elif op=="//":
    result = n1//n2
  elif op=="^":
    result = n1**n2%
  print(f"{n1} {op} {n2} = {result}")


print("Thank you!")
```

### lambda function:
* it is one linear
* it is defined when we want to use it only once
* it is an anonymous function i,e.without name
* syntax:lambda parameters_list:return_statements

```python
def add(a,b):
    return a+b
print(add(10,20))
```

```python
add2=lambda a,b:a+b
print(add2(10,20))
```

```python
# sort the following list of tuples as per the 2nd vaues in the tuples
nums=[(12,23),(45,13),(20,25),(18,8)]
nums.sort()#by default list is sorted by 1st values in the typles
print(nums)
nums.sort(key=lambda x:x[1])#list is sorted by 2nd value in the tuples.
print(nums)
```

```python
# sorting employees
employees=[('amit',10000),('sumit',8500),('rakshit',6000)]
#sort employees according to their names
employees.sort()
print(employees)
#sort employees according to their salaries
employees.sort(key=lambda t:t[1])
print(employees)
```

### Random library:

```python
import random

print(random.randint(11,20)) # generate random integer between given ranne(both inclusive)
print(random.random()) # random float value between 0-1(0 included,1 not included)
```

```python
random.seed(12)
#generate random 20 integer between 1 and 70
for i in range(20):
    print(random.randint(1,70),end=" ")
sr_num=[]
for i in range(70):
    sr_num.append(i+1)
#print(sr_num)
print()
random.choices(sr_num,k=20)
```

```python
# random selection without repetation:
sr_num=[]
for i in range(70):
    sr_num.append(i+1)
#method1:
print(random.sample(sr_num,k=20))
#method2:
random.shuffle(sr_num)#original list gets updated  y shuffling the elements randomly
print(sr_num[:20])
```
