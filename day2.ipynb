### break and continue

```python
#print all the number between 1 and 30
#skip all the numbers divisible by 3
# stop printing when the number is greater than 20
for i in range(1,31):
    if i%3==0:
        continue #skip next part of the block and continue to next value in the loop
    if i>20:
        break # exiting from loop
    print(i,end=" ")
```

### take a number input from the user and print the sum of all its digit

```python
n=int(input("enter a number:"))
sum=0
while n>0:
    digit=n%10
    sum=sum+digit
    n=n//10
print("sum of digits:",sum)

```

### preprocessing:given any string,remove all the digits and special characters except space from the string i.e.keep only alphabets and spaces.

```python
msg = input("enter any message:")
preprocessed_msg =""
for ch in msg:
    ascii_val = ord(ch)
    if ascii_val==32 or 65<=ascii_val<=90 or 97<=ascii_val<=122:
        preprocessed_msg += ch
print(f"original message:{msg}")
print("preprocessed message:",preprocessed_msg)
```
