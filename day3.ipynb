```python
def print_table(n):
    for i in range(1,11):
        print(f"{n} * {i} ={n*i}")
    print("#"*10) #string printed multiple times

print_table(5)
print_table(6)
print_table(7)
```

### write a function to calculate the average of all the numbers in a given list.

```python
def average(nums):
    total=sum(nums)
    count=len(nums)
    return total/count
print(average([10,20,30,40,50]))
print(average([1,2,3,4,5]))
```

### standard deviation

```python
def average(nums):
    n=len(nums)
    mean=sum(nums)/n
    variance=sum((x-mean)**2 for x in nums)/n
    standard_deviation=variance**0.5
    print(standard_deviation)
average([1,2,3,4,5,6])
```

```python
def average(nums):
    return sum(nums)/len(nums)

def std(nums):
    mu=average(nums)
    total=0
    for x in nums:
        total += (x-mu)**2
    sd=(total/len(nums))**0.5
    return sd
print(std([8,5,7,6,8,6]))
print(std([1,2,3,4,5,6]))
```

```python
def print_pattern(n):
    for row in range(n):
        for sp in range(n-row-1):
            print(" ", end=" ")
        for str in range(2*row+1):
            print(f"{str+1}", end=" ")
        print()

    for row in range(n-1):
        for sp in range(row+1):
            print(" ", end=" ")
        for str in range(2*(n-row-2)+1):
            print(f"{str+1}", end=" ")
        print()

print_pattern(5)
```

### slicing of string:fetching some part of the string
parameters in slicing:[start=0:end=len(str):step=1]
start("first 10 characters:",text[0:10])

```python
text="abcdefghijklmnopqrstuvwxyz"
print(len(text))
print("first 10 characters:",text[0:10])
print("Character from 6 to 10 index:",text[6:11])
print("Last 10 characters:",text[len(text)-10:len(text)])
print("last 10 characters:",text[-10:-1]+text[-1])
print("last 10 characters:",text[::2])
print("Slicing with step 5:",text[::5])
print("reversed:",text[-1:-len(text)-1:-1])
print("reversed:",text[::-1])
```
